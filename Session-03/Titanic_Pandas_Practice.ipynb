## Import Libraries & Load Dataset

```python
import pandas as pd
import numpy as np

titanic = pd.read_csv("titanic (1).csv")
titanic.head()
```

## Explore Dataset

```python
print("Shape:")
print(titanic.shape)

print("\nInfo:")
titanic.info()

print("\nDescribe:")
print(titanic.describe(include="all"))

print("\nMissing Values:")
print(titanic.isnull().sum())
```

## Select Columns

```python
selected_df = titanic[["Survived", "Pclass", "Sex", "Age", "Fare"]]

selected_df.head()
```

## Filter Female, First Class, Survived

```python
female_first_survived = titanic[
    (titanic["Sex"] == "female") &
    (titanic["Pclass"] == 1) &
    (titanic["Survived"] == 1)
]

female_first_survived
```

## Sort by Fare

```python
top_fare = titanic.sort_values(by="Fare", ascending=False)

top_fare[["Name", "Fare"]].head(10)
```

## Create AgeGroup Column

```python
def age_group(age):
    if pd.isna(age):
        return "Unknown"
    elif age < 13:
        return "Child"
    elif age <= 19:
        return "Teen"
    elif age <= 59:
        return "Adult"
    else:
        return "Senior"

titanic["AgeGroup"] = titanic["Age"].apply(age_group)

titanic[["Age", "AgeGroup"]].head(10)
```

## Handle Missing Values

## Check Missing Values

```python
titanic.isnull().sum()
```

## Fill Age with Median

```python
titanic["Age"] = titanic["Age"].fillna(titanic["Age"].median())
```

## Fill Embarked with Mode

```python
titanic["Embarked"] = titanic["Embarked"].fillna(
    titanic["Embarked"].mode()[0]
)
```

## Fill Cabin

```python
titanic["Cabin"] = titanic["Cabin"].fillna("Unknown")
```

## Verify

```python
titanic.isnull().sum()
```

## Remove Duplicates

```python
duplicates = titanic.duplicated().sum()

print("Duplicates:", duplicates)
```

## Group & Aggregate

## Survival Rate by Sex

```python
survival_by_sex = titanic.groupby("Sex")["Survived"].mean()

print(survival_by_sex)
```

## Survival Rate by Passenger Class

```python
survival_by_class = titanic.groupby("Pclass")["Survived"].mean()

print(survival_by_class)
```

## Pivot Table

```python
pivot = titanic.pivot_table(
    values="Survived",
    index="Pclass",
    columns="Sex",
    aggfunc="mean"
)

pivot
```

## Extract Passenger Title

```python
titanic["Title"] = titanic["Name"].str.extract(r',\s*([^\.]+)\.')

titanic["Title"].value_counts()
```

## Survival Rate by Title

```python
title_survival = titanic.groupby("Title")["Survived"].mean().sort_values(ascending=False)

title_survival
```

## Import Pandas

```python
import pandas as pd
```

## Load Dataset

```python
df = pd.read_csv("medicine_inventory.csv")
df
```

## Display First and Last Rows

```python
print(df.head())
print(df.tail(2))
```

## Check Missing Values

```python
print(df[["Price", "Quantity"]].isnull().sum())
```

## Fill Missing Values

```python
df["Price"] = df["Price"].fillna(df["Price"].mean())
df["Quantity"] = df["Quantity"].fillna(1)
df
```

## Create Inventory Value

```python
df["Inventory_Value"] = df["Price"] * df["Quantity"]
```

## Boolean Filtering

```python
well_stocked = df[df["Quantity"] >= 15]
well_stocked
```

## loc Selection

```python
df.loc[5:10, ["Medicine", "Inventory_Value"]]
```

## Statistical Summary

```python
df["Inventory_Value"].describe()
```

## Save Cleaned Dataset

```python
df.to_csv("cleaned_medicine_inventory.csv", index=False)
print("Cleaned dataset saved successfully.")
```

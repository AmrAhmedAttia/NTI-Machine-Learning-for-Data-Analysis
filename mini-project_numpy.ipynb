### Import & Create Data

```python
import numpy as np
fitness_data = np.array([
    [5000, 200, 6],
    [7000, 250, 7],
    [6500, 230, 6],
    [8000, 300, 8],
    [9000, 320, 7],
    [7500, 270, 6],
    [12000, 400, 5]])
    
```

### Explore the Array

```python
print("Number of Dimensions:", fitness_data.ndim)
print("Shape:", fitness_data.shape)
print("Total Number of Elements:", fitness_data.size)
```

### Slice the Data

```python
steps = fitness_data[:, 0]
calories = fitness_data[:, 1]
sleep = fitness_data[:, 2]

print("Steps:")
print(steps)

print("\nCalories:")
print(calories)

print("\nSleep Hours:")
print(sleep)

print("\nFirst 3 Days:")
print(fitness_data[:3])

print("\nSteps and Calories:")
print(fitness_data[:, :2])
```

### Reshape the Data

```python
steps_column = steps.reshape(7, 1)

print("Steps Reshaped to (7,1):")
print(steps_column)

reshaped_data = fitness_data.reshape(3, 7)

print("\nDataset Reshaped to (3,7):")
print(reshaped_data)
```

### Filter the Data

```python
print("Steps > 8000")
print(fitness_data[fitness_data[:, 0] > 8000])

print("\nSleep < 6")
print(fitness_data[fitness_data[:, 2] < 6])

print("\nCalories > 300")
print(fitness_data[fitness_data[:, 1] > 300])
```

### Statistical Functions

```python
print("Average Steps:", np.mean(steps))
print("Maximum Steps:", np.max(steps))
print("Minimum Sleep Hours:", np.min(sleep))

print("Standard Deviation of Calories:", np.std(calories))

print("Range of Steps:", np.ptp(steps))
```

### Insights

## Insights

### Patterns
- Calories burned increase as the number of steps increases.
- Most days have between 6 and 7 hours of sleep.

### Unusual Day
- The day with 12,000 steps and 400 calories burned is the most active.
- It also has the lowest sleep duration (5 hours).

### What I Learned
- More physical activity usually burns more calories.
- High activity does not always mean enough sleep.

### Scouting the Academy's Next Star - NumPy Mini Project

### Import Libraries

```python
import numpy as np
```

### Load Dataset

```python
players = np.array([
    "Adam", "Mostafa", "Yara", "Hana", "Ziad", "Malak",
    "Omar", "Farida", "Karim", "Nada", "Tarek", "Reem"
])

positions = np.array([
    "Forward", "Defender", "Midfielder", "Forward", "Goalkeeper", "Defender",
    "Midfielder", "Forward", "Defender", "Midfielder", "Forward", "Goalkeeper"
])

performance = np.array([
    [28.5, 82, 75, 88],
    [26.0, 90, 88, 70],
    [30.2, 78, 60, 92],
    [27.8, 85, 70, 80],
    [24.5, 95, 92, 65],
    [25.9, 70, 65, 75],
    [29.1, 80, 55, 90],
    [31.0, 60, 50, 95],
    [23.8, 88, 90, 60],
    [28.9, 83, 68, 85],
    [27.2, 76, 72, 78],
    [22.9, 93, 95, 58],
])

print(players.shape)
print(positions.shape)
print(performance.shape)
```

### Explore Dataset

```python
print(performance[0])
print(performance[5])
print(performance[11])
```

### Clean Data

```python
performance_clean = performance.copy()

performance_clean[:, 2] = np.clip(
    performance_clean[:, 2],
    None,
    90
)

print(performance_clean)
```

### Calculate Talent Score

```python
weights = np.array([1.5, 0.7, 0.5, 1.3])

weighted_performance = performance_clean * weights

talent_score = np.sum(weighted_performance, axis=1)

print(talent_score)
```

## Create Structured Array

```python
data = [
    (players[i], positions[i], talent_score[i])
    for i in range(len(players))
]

Squad = np.array(
    data,
    dtype=[
        ("name", "U10"),
        ("position", "U12"),
        ("score", "f4")
    ]
)

print(Squad.dtype.names)
print(Squad.shape)
print(Squad["name"])
print(Squad["score"])
```

## Filter Goalkeepers

```python
goalkeepers = Squad[Squad["position"] == "Goalkeeper"]

print(goalkeepers)
```

## Sort Players by Score

```python
sorted_Squad = np.sort(
    Squad,
    order="score"
)[::-1]

print(sorted_Squad)
```

## Top 3 Players

```python
record_Squad = sorted_Squad.view(np.recarray)

print(record_Squad.name[:3])
print(record_Squad.score[:3])
```

## Select Top 6 Players

```python
final_squad = sorted_Squad[:6]

print(final_squad)
```

## Apply Goalkeeper Rule

```python
has_goalkeeper = np.any(
    final_squad["position"] == "Goalkeeper"
)

if not has_goalkeeper:
    removed_player = final_squad[-1]

    best_goalkeeper = np.sort(
        goalkeepers,
        order="score"
    )[-1]

    final_squad[-1] = best_goalkeeper

    print("Removed Player:")
    print(removed_player)

    print("Added Goalkeeper:")
    print(best_goalkeeper)

print(final_squad)
```

# Final Report

The final squad was selected based on the highest Talent Scores while satisfying the goalkeeper requirement.

The selected players are:
